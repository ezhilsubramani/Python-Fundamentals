# **List** (Creation, Modification and Access):

List Creation:

```python
 #Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28]
 age_list=[28,29,30,31,32]
 age_list
```

```python
 #Create a list named name_list with five string elements.
 name_list= ['Ezhil','Suva','Kavi','Jeeva','Shalini']
 name_list
```

List Operations / Modifications:

```python
#Append the string "Yazhini" to name_list.
name_list.append("Yazhini")
print(name_list)
```

```python
#Insert the element 30 at index 2 in age_list.
age_list[2]=35
print(age_list)
age_list.insert(2,30)
print(age_list)
```

```python
#Remove the string "Yazhini" from name_list.
print(name_list)
name_list.append("Yazhini")
name_list.remove("Yazhini")
print(name_list)
```

```python
#Pop the last element from age_list.
age_list.pop()
print(age_list)
```

```python
#Extend the age_list with additional ages [29, 30, 26].
age2=[29,30,26]
age_list += age2
print(age_list)
```

```python
#Sort age_list in descending order.
age_list.sort(reverse=True)
print(age_list)
```

```python
#Find Max age, Min age and sum of all ages from age_list.
max(age_list)
```

```python
min(age_list)
```

```python
sum(age_list)
```

```python
#Create a set called my_set with following values:
my_list=['a','e','i','o','u','a','a','i']
my_set=set(my_list)
print(my_set)
print(type(my_set))
#Explanation= set never contains duplicate value
```

```python
#Create two sets:
set1={1, 3, 5, 7, 9}
set2= {2, 3, 5, 8, 10}
```

```python
#Compute and print the union and intersection of set1 and set2.
#union
all_set=set1 | set2
all_set
```
