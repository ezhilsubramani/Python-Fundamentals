# **While Loop**

**Problem Statement:** Create a Python program that implements a simple number
guessing game using a while loop. The program should make use of control
statements such as else, break, and continue.

```python
#Set Up the Game:
import random
# Generate random number between 1 and 10
secret_number = random.randint(1, 10)
#Prompt the User:
attempts=3
#Implement the Guessing Logic:
while attempts>0:
 number=int(input("Guess the Number in the range(1,10): "))
 if number==secret_number:
  print("Congradulations! Your guess is correct.")
  break
 if number>10:
  print("Your guess is out of range. Please guess a number between 1 and 10.")
  continue
 if number<secret_number:
    print("Too low. Try again.")
 elif number>secret_number:
    print("Too high. Try again.")
 attempts-=1
 print('remaining attempts: ',attempts)
else:
  print("Too many attempts sorry you are out of the game")
  print("The secret number was",secret_number)
```

# **For Loop**

**Problem Statement:** Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range
function.

```python
#Prompt user for Input. Ask the user to enter a number for which they want to generate a multiplication table.
number=int(input("Enter the number for you want a multiplication table: "))
#Generate the Multiplication Table:
for i in range(1,11):
  result=number*i
  print(number,'X',i,'=',result)
```

**Problem Statement**: Create a Python program that calculates the Body Mass Index(BMI).
              Hint: BMI = weight (kg) / [height (m)]²

```python
def BMI(weight, height):
    return weight / (height ** 2)

for i in range(4):

    weight = int(input("Enter your weight in kg: "))
    height = float(input("Enter your height in m: "))

    bmi = BMI(weight, height)

    if bmi < 18.5:
        print("Your BMI is:", round(bmi, 2), "Underweight")

    elif 18.5 <= bmi <= 24.9:
        print("Your BMI is:", round(bmi, 2), "Healthy Weight")

    elif 25 <= bmi <= 29.9:
        print("Your BMI is:", round(bmi, 2), "Overweight")

    else:
        print("Your BMI is:", round(bmi, 2), "Obese")
```
